````python
import torch

# ==========================================
# 1. 定义状态转移矩阵 (Transition Matrix)
# ==========================================
# 对应图中的 T 矩阵。
# 矩阵的行代表“当前状态 (Current State)”，列代表“下一个状态 (Next State)”。
# 索引 0 代表 s0 (study)，索引 1 代表 s1 (sleep)
T = torch.tensor([[0.4, 0.6], # 当前在 s0: 0.4的概率留在s0，0.6的概率去s1
                  [0.8, 0.2]]) # 当前在 s1: 0.8的概率去s0，0.2的概率留在s1

# ==========================================
# 2. 计算 n 步转移矩阵 (n-step Transition Matrix)
# ==========================================
# 矩阵乘法具有结合律，T^n 代表连续进行 n 次状态转移后的概率矩阵。
# 矩阵中的 T^n[i][j] 表示：从状态 i 出发，经过 n 步后刚好到达状态 j 的概率。
T_2 = torch.matrix_power(T, 2)  # 2步后的转移概率
T_5 = torch.matrix_power(T, 5)  # 5步后的转移概率
T_10 = torch.matrix_power(T, 10) # 10步后的转移概率
T_15 = torch.matrix_power(T, 15) # 15步后的转移概率
T_20 = torch.matrix_power(T, 20) # 20步后的转移概率

# 打印不同步数下的转移矩阵，观察其收敛趋势
print("Transition probability after 2 steps:\n{}".format(T_2))
print("Transition probability after 5 steps:\n{}".format(T_5))
print("Transition probability after 10 steps:\n{}".format(T_10))
print("Transition probability after 15 steps:\n{}".format(T_15))
print("Transition probability after 20 steps:\n{}".format(T_20))

# ==========================================
# 3. 定义初始状态分布 (Initial State Distribution)
# ==========================================
# 一个 1x2 的行向量，表示系统在第 0 时刻处于各个状态的概率。
# 0.7 表示有 70% 的概率从 s0 (study) 开始
# 0.3 表示有 30% 的概率从 s1 (sleep) 开始
v = torch.tensor([[0.7, 0.3]])

# ==========================================
# 4. 计算 n 步后的状态分布 (n-step State Distribution)
# ==========================================
# 状态分布的演化公式：v_next = v_current * T
# 这是一个全概率的计算过程：下一时刻在 j 的概率 = 求和(当前在 i 的概率 * 从 i 到 j 的概率)
v_1 = torch.mm(v, T)      # 第 1 步后的状态分布
v_2 = torch.mm(v, T_2)    # 第 2 步后的状态分布 (等同于 v_1 * T)
v_5 = torch.mm(v, T_5)    # 第 5 步后的状态分布
v_10 = torch.mm(v, T_10)  # 第 10 步后的状态分布
v_15 = torch.mm(v, T_15)  # 第 15 步后的状态分布
v_20 = torch.mm(v, T_20)  # 第 20 步后的状态分布

# 打印不同时刻下的状态分布，观察其收敛到稳定分布的过程
print("Distribution of states after 1 step:\n{}".format(v_1))
print("Distribution of states after 2 steps:\n{}".format(v_2))
print("Distribution of states after 5 steps:\n{}".format(v_5))
print("Distribution of states after 10 steps:\n{}".format(v_10))
print("Distribution of states after 15 steps:\n{}".format(v_15))
print("Distribution of states after 20 steps:\n{}".format(v_20))
````

这段代码在数学和机器学习层面模拟了一个简单的**马尔可夫链（Markov Chain）**。


下面是代码的逐块详细解读：

### 1. 转移矩阵的幂运算
```python
T = torch.tensor([[0.4, 0.6],
                  [0.8, 0.2]])

T_2 = torch.matrix_power(T, 2)
# ... (略)
print("Transition probability after 2 steps:\n{}".format(T_2))
```
*   **`T` 代表什么？** 
    这就是你图片中的矩阵 $T$。它是一个**一步转移概率矩阵**。
    *   `T[0][0] = 0.4`：代表当前处于状态 $s_0$ (study)，下一步仍然留在 $s_0$ 的概率是 0.4（对应图中的自环）。
    *   `T[0][1] = 0.6`：代表当前处于状态 $s_0$ (study)，下一步切换到 $s_1$ (sleep) 的概率是 0.6。
*   **`T_2`, `T_5` 等代表什么？**
    通过矩阵自乘（求幂），它计算了**多步转移概率矩阵**。
    *   `T_2`：代表从任意一个状态出发，经过 **2步** 后到达其他状态的概率。
    *   `T_10`：代表经过 **10步** 后的转移概率。

### 2. 状态分布的演化
```python
v = torch.tensor([[0.7, 0.3]])

v_1 = torch.mm(v, T)
v_2 = torch.mm(v, T_2)
# ... (略)
print("Distribution of states after 1 step:\n{}".format(v_1))
```
*   **`v` 代表什么？**
    这是一个**初始状态分布向量**。`[[0.7, 0.3]]` 意味着系统在第 0 步时，有 70% 的概率在 $s_0$ (study)，有 30% 的概率在 $s_1$ (sleep)。
*   **`v_1`, `v_2` 等代表什么？**
    通过状态分布向量与转移矩阵相乘（`torch.mm` 即矩阵乘法），它计算了系统经过若干步后的**具体状态分布**。
    *   例如 `v_1 = torch.mm(v, T)` 就是在算：第一步之后，系统处于 study 和 sleep 的概率分别是多少。

### 3. 这段代码的“终极目的”是什么？
当你运行这段代码时，你会发现一个非常神奇的现象：

*   随着步数从 1 增加到 20（即看 `T_10`, `T_20` 或 `v_10`, `v_20` 的输出）：
    1.  多步转移矩阵 $T$ 的每一行都会变得**一模一样**。
    2.  无论你初始分布 `v` 是什么（哪怕你改成 `[[0.0, 1.0]]` 全在 sleep），随着步数增多，最终计算出的状态分布都会**收敛到同一个数值**（大约是 `[[0.5714, 0.4286]]`）。

这直观地证明了：**一个正常且连通的马尔可夫链，存在一个与初始状态无关的稳定概率分布（Stationary Distribution）。** 这一概念是后面学习强化学习（如 MDP、策略评估）的最基础数学根基。

用 $V$ (状态分布向量) 乘以 $T$ (转移矩阵)，即 $V \times T$，在数学上叫做 **“全概率公式”的矩阵化表达**。

它的核心含义是：**计算系统在“下一步”处于各个状态的“总概率”（即预测未来的状态分布）。**

为了让彻底理解，我们不谈抽象的矩阵，直接用图中的“学习 ($s_0$)”和“睡觉 ($s_1$)”两个状态，结合具体的概率来一步步拆解。

### 1. 初始状态：你现在在哪里？
`v = [[0.7, 0.3]]`
这句话的意思是：在当前时刻（第 0 步），你有 **70% (0.7)** 的概率正在“学习”，有 **30% (0.3)** 的概率正在“睡觉”。

### 2. 目标：你想预测什么？
你想预测：**1步之后（第 1 步）**，你处于“学习”和“睡觉”的概率分别是多少？

### 3. 为什么需要相乘？（分步推理）
因为未来的状态取决于“你当前在哪”以及“转移的概率”。

假设我们要算**1步之后，你正在“睡觉”的概率**，一共有两条路径：
*   **路径 1**：你现在正在“学习”，然后**保持**学习，然后**转移**到了“睡觉”。
    *   概率 = (当前在学习的概率) $\times$ (从学习转移到睡觉的概率) = $0.7 \times 0.6 = 0.42$
*   **路径 2**：你现在正在“睡觉”，然后**保持**睡觉。
    *   概率 = (当前在睡觉的概率) $\times$ (从睡觉转移到睡觉的概率) = $0.3 \times 0.2 = 0.06$

因为这两条路径都能让你在 1 步之后处于“睡觉”状态，所以必须把它们加起来：
*   1步之后在“睡觉”的总概率 = $0.42 + 0.06 = \mathbf{0.48}$

同理，我们算一下**1步之后，你正在“学习”的概率**：
*   **路径 A**：从“学习”保持在“学习”：$0.7 \times 0.4 = 0.28$
*   **路径 B**：从“睡觉”转移到“学习”：$0.3 \times 0.8 = 0.24$
*   1步之后在“学习”的总概率 = $0.28 + 0.24 = \mathbf{0.52}$

所以，1步之后的状态分布就是 `[[0.52, 0.48]]`。

### 4. 矩阵乘法 $V \times T$ 就是把这些计算打包在一起
你看上面的计算过程，是不是完全吻合矩阵乘法的规则？

$$
\begin{aligned}
\text{新状态向量 } V_1 &= V \times T \\
&= [0.7, \quad 0.3] \times \begin{bmatrix} 0.4 & 0.6 \\ 0.8 & 0.2 \end{bmatrix} \\
&= [ (0.7 \times 0.4 + 0.3 \times 0.8), \quad (0.7 \times 0.6 + 0.3 \times 0.2) ] \\
&= [ (0.28 + 0.24), \quad (0.42 + 0.06) ] \\
&= [ 0.52, \quad 0.48 ]
\end{aligned}
$$

### 总结
*   **$V$**：代表当前时刻系统的状态分布（你在哪，概率多大）。
*   **$T$**：代表状态变化的规律（从这去哪，概率多大）。
*   **$V \times T$**：代表利用当前的分布和变化的规律，推演出**下一时刻**系统的状态分布。

在强化学习中，Agent（智能体）每走一步，环境的状态就会根据转移概率发生变化，我们就是不断地用 `当前分布 * T` 来模拟和预测 Agent 在环境中的演化过程。

理解了这一步的矩阵运算，接下来代码里的 `T` 的幂次方（$T$ 的 $n$ 次方）其实就是把这种“预测”一次性往前推了 $n$ 步。
