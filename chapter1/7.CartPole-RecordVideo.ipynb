```python
# 导入 gymnasium 库，并用 gym 作为别名，方便后续调用
import gymnasium as gym
# 从 gymnasium.wrappers 中导入 RecordVideo 包装器，用于自动录制环境视频
from gymnasium.wrappers import RecordVideo

# 定义视频保存的文件夹路径
video_dir = "./cartpole_video/"

# 创建 CartPole-v1 环境，render_mode="rgb_array" 表示以图像数组形式渲染，这样才能录制视频
env = gym.make("CartPole-v1", render_mode="rgb_array")

# 用 RecordVideo 包装环境，使其具备自动录制视频的功能
env = RecordVideo(
    env,
    video_folder=video_dir,       # 视频保存的目录
    name_prefix="cartpole",       # 视频文件名的前缀
    episode_trigger=lambda ep: ep % 100 == 0  # 每 100 个 episode 录制一次视频
)

# 重置环境，获取初始观测值 obs 和额外信息 info
obs, info = env.reset()
# 初始化终止标志，terminated 表示任务是否自然结束（如杆子倒下）
terminated = False
# 初始化截断标志，truncated 表示是否因达到最大步数而结束
truncated = False

# 当任务既没有自然结束也没有被截断时，持续执行
while not (terminated or truncated):
    # 从动作空间中随机采样一个动作
    action = env.action_space.sample()
    # 执行动作，获取新的观测值、奖励、终止标志、截断标志和额外信息
    obs, reward, terminated, truncated, info = env.step(action)

# 第一个 episode 结束后打印提示
print("第一个 episode 结束")

# 设置总共要跑的 episode 数量
n_episode = 10000
# 用于存储每个 episode 的总奖励
total_rewards = []

# 循环跑 n_episode 个 episode
for episode in range(n_episode):
    # 重置环境，开始新的 episode
    obs, info = env.reset()
    # 初始化当前 episode 的累计奖励
    total_reward = 0.0
    # 重置终止和截断标志
    terminated = False
    truncated = False

    # 当前 episode 未结束时持续执行
    while not (terminated or truncated):
        # 随机采样一个动作
        action = env.action_space.sample()
        # 执行动作，获取反馈信息
        obs, reward, terminated, truncated, info = env.step(action)
        # 累加当前 episode 的奖励
        total_reward += reward

    # 将当前 episode 的总奖励存入列表
    total_rewards.append(total_reward)

# 计算并打印 10000 个 episode 的平均总奖励
print(
    f"Average total reward over {n_episode} episodes: "
    f"{sum(total_rewards) / n_episode:.2f}"
)

# 关闭环境，释放资源
env.close()
```
