```python
import gymnasium as gym

# 创建环境
env = gym.make("CartPole-v1")

# 测试回合数
n_test_episodes = 10
# 仅在第一个回合的前 20 步打印详细中文信息，避免刷屏
n_print_steps = 20

for episode in range(n_test_episodes):
    # 重置环境，获取初始状态
    obs, info = env.reset()
    terminated = False
    truncated = False
    step_count = 0
    total_reward = 0.0

    print(f"--- Episode {episode} 开始 ---")
    
    while not (terminated or truncated):
        # 随机选择一个动作 (0:向左, 1:向右)
        action = env.action_space.sample()
        obs, reward, terminated, truncated, info = env.step(action)
        total_reward += reward
        step_count += 1
        
        # 仅在第一个 episode 的前 n_print_steps 步打印格式化中文信息
        if episode == 0 and step_count <= n_print_steps:
            cart_pos, cart_vel, pole_angle, pole_vel = obs
            
            # 判断小车是否即将超出轨道范围 (-2.4 到 2.4)
            cart_status = "越界终止" if abs(cart_pos) > 2.4 else "正常"
            # 判断杆子是否即将倾斜过度 (-0.209 到 0.209 弧度)
            pole_status = "倾斜终止" if abs(pole_angle) > 0.209 else "正常"

            print(f"Step {step_count:>2} | "
                  f"位置: {cart_pos:+.4f} ({cart_status}) | "
                  f"速度: {cart_vel:+.4f} | "
                  f"角度: {pole_angle:+.4f} ({pole_status}) | "
                  f"角速度: {pole_vel:+.4f}")

    print(f"Episode {episode} 结束 | 存活步数: {step_count} | 总奖励: {total_reward}\n")

env.close()
```
